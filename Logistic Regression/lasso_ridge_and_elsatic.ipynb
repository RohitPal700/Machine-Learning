```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
df=sns.load_dataset('mpg')
```

```python
df.head()
```

```python
df.drop("name",axis=1,inplace=True)
```

```python
df.isna().sum()
```

```python
df['horsepower'].median()
```

```python
df['horsepower']=df['horsepower'].fillna(df['horsepower'].median())
```

```python
df.dtypes
```

```python
df['origin']
```

```python
df['origin'].value_counts()
```

```python
df['origin']=df['origin'].map({"usa":1,"japan":2,"europe":3})
```

```python
df['origin'] =df['origin'].astype(int)
```

```python
df.dtypes
```

```python
#separate x and y
X=df.drop('mpg',axis=1)
y=df['mpg']
```

```python
X
```

```python
y
```

```python
#train test split
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=1)
```

```python
X_train.shape,X_test.shape
```

```python
# smiple linea regression model 
from sklearn.linear_model import LinearRegression
regression_model=LinearRegression()
```

```python
regression_model
```

```python
regression_model.fit(X_train,y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f'The coefficient for {col_name} is {regression_model.coef_[i]}')
```

```python
#coefficient are relatively smaller , if one independent variable changes 
#slightly there will be not much difference in prediction 
#this is sometime is called smoother model 
```

```python
from sklearn.metrics import r2_score

y_pred_linear =regression_model.predict(X_test)
r2_linear= r2_score(y_test,y_pred_linear)

print(f"R square of linear regression {r2_linear}")
```

```python
# Ridge regression 
from sklearn.linear_model import Ridge
ridge_regression_model = Ridge(alpha=0.1)

ridge_regression_model.fit(X_train,y_train)


for i, col_name in enumerate(X_train.columns):
    print(f"The coefficient for {col_name} is {ridge_regression_model.coef_[i]}")
```

```python
# For Ridge Regression evaluation 
y_pred_ridge =ridge_regression_model.predict(X_test)
r2_ridge =r2_score(y_test,y_pred_ridge)

print(f"R-suared score for ridge Regression: {r2_ridge}")
```

```python
# We dont see much variaation in coeff of rifge regression as compares to linear regression 
```

```python
from sklearn.linear_model import Lasso 
Lasso_ression_model= Lasso(alpha=0.5)
Lasso_ression_model.fit(X_train,y_train)

for i,col_name in enumerate (X_train.columns):
    print(f"The coefficient for {col_name} is {Lasso_ression_model.coef_[i]}")
```

```python
#3 feature coefficient are 0, Lasso helps in feature selction 
```

```python
y_pred_lasso =Lasso_ression_model.predict(X_test)
r2_lasso =r2_score(y_test,y_pred_lasso)
print(f"R-squared score for Lasso Regression : {r2_lasso}")
```

```python
from sklearn.linear_model import ElasticNet

ElasticNet_regression_model=ElasticNet(alpha=1,l1_ratio=0.5)
ElasticNet_regression_model.fit(X_train,y_train)
```

```python
for i,col_name in enumerate(X_train.columns):
    print(f"The coefficient for {col_name} is {ElasticNet_regression_model.coef_[i]}")
```

```python
y_pred_elastic_net =ElasticNet_regression_model.predict(X_test)

r2_elastic_net = r2_score(y_test,y_pred_elastic_net)

print(f"R-squares score for Elastic Net Regression : {r2_elastic_net}")
```
