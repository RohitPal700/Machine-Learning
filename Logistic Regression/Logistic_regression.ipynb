```python
import pandas as pd
import numpy as np 
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
```

```python
# For Binary Classification 
df_final=df[df['target']!=2]
```

```python
df_final
```

```python
df_final.target.unique()
```

```python
X=df_final.iloc[:,:-1]
y=df_final.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=1)
```

```python
X_train.shape
```

```python
# Model Training 
```

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression()
```

```python
classifier
```

```python
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
```

```python
classifier.predict_proba(X_test) # corresponding to higher probability class is predicted , cutoff 0.5
```

```python
# Evalution metrics

from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))

# Frist two rows , 0 and  1>> precision, recall, f1-score, and support = No od data points of respective class 
#macro avg >> for each class take simple avg >> avg prescision of class 0 and class 1 will be macro avg
#similarly for recall and ,f1 score ,support 

#weighted avg is used in case of class imbalance , it gives importance to performance on dominant class
```

```python
from sklearn.metrics import roc_curve,auc
import matplotlib.pyplot as plt

#Get predict probabilities for class 1 
classifier.predict_proba(X_test)[:,1] # probability for one class


```

```python
y_pred_proba=classifier.predict_proba(X_test)[:,1]
```

```python
# Reciever operating characteristics curve 
#inputs: y_test , y_pred_proba . It will return TPR,FPR with different cutoff of probabilities 
#how well your model is able to distinguish b/w two classes 
fpr , tpr , thresholds = roc_curve(y_test,y_pred_proba) 
```

```python
fpr
```

```python
tpr
```

```python
thresholds
```

```python
#calculate auc score 
roc_auc=auc(fpr,tpr)
roc_auc
```

```python
# Plot the ROC-AUC curve to evaluate the classification model's performance
plt.figure(figsize=(8, 6))

# Plot the ROC curve using False Positive Rate (fpr) and True Positive Rate (tpr)
# Set the curve color to dark orange and its line width to 2
# Display the ROC-AUC score in the legend
plt.plot(fpr, tpr, color='darkorange', linewidth=2,
         label="ROC curve (area = %0.2f)" % roc_auc)

# Plot the diagonal reference line representing a random classifier
# The line runs from (0, 0) to (1, 1), with a navy color and dashed style
plt.plot([0, 1], [0, 1], color='navy', linewidth=2, linestyle='--')

# Set the minimum and maximum values of the X-axis (False Positive Rate)
plt.xlim([0.0, 1.0])

# Set the minimum and maximum values of the Y-axis (True Positive Rate)
plt.ylim([0.0, 1.05])

# Label the X-axis
plt.xlabel("False Positive Rate")

# Label the Y-axis
plt.ylabel("True Positive Rate")

# Set the title of the graph
plt.title("Receiver Operating Characteristic (ROC) Curve")

# Display the legend in the lower-right corner of the graph
plt.legend(loc='lower right')

# Display the completed ROC curve
plt.show()
```
