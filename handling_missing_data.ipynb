```python
import numpy as np 
import pandas as pd 

np.random.seed(1) # if we fix seed value then same data will be genrate (for reproducibility)
```

```python
no_sample = 1000

no_class_0 = 900
no_class_1 = 100
```

```python
class_0 = {'feature1': np.random.normal(0,1,no_class_0), 
           'feature2': np.random.normal(0,1,no_class_0),
           'target':[0]*no_class_0}
class_1 ={'feature1': np.random.normal(0,1,no_class_1), 
           'feature2': np.random.normal(0,1,no_class_1),
           'target':[1]*no_class_1}
```

```python
class_0=pd.DataFrame(class_0)
class_1=pd.DataFrame(class_1)
```

```python
class_0
```

```python
class_1
```

```python
#now cobile both class class_0 and class_1
df=pd.concat([class_0,class_1]).reset_index(drop=True )
```

```python
df
```

```python
df.target.value_counts()
```

```python
#upsampling

df_minority = df[df.target == 1]
df_majority = df[df.target == 0]
```

```python
df_minority
```

```python
from sklearn.utils import resample
```

```python
df_majority_upsampled= resample(df_minority , replace= True,n_samples= len(df_majority),random_state=1)
```

```python
df_majority_upsampled
```

```python
df_upsampled = pd.concat([df_majority,df_majority_upsampled])

df_upsampled
```

```python
#downsampling 

df_majority_downsample= resample(df_majority,replace=False,n_samples=len(df_minority),random_state=1)
```

```python
df_majority_downsample
```

```python
df_downsampled= pd.concat([df_minority,df_majority_downsample])
```

```python
df_downsampled
```
